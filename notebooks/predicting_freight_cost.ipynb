```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sqlite3
```

# Predicting Freight Cost

**Objective:** Predict freight cost for a vendor invoice using quantity and dollars, to improve cost forecasting, budgeting, and vendor negotiation.

- Freight is a non-trivial component of landed cost.
- Poor freight estimates distort margin and inventory planning.
- Automating freight estimation helps procurement teams forecast true cost before invoice arrival.


```python
conn = sqlite3.connect("../data/inventory.db")
```

```python
tables = pd.read_sql_query("select name from sqlite_master where type = 'table'", conn)
tables
```

```python
for table in tables["name"]:
    print(f"{"==" * 10} table: {table} {"==" * 10}")
    df = pd.read_sql_query(f"select * from {table} limit 5", conn)
    display(df)
```

```python
vendor_df = pd.read_sql_query("select * from vendor_invoice", conn)
vendor_df.head()
```

```python
vendor_df[["Quantity", "Freight", "Dollars"]].corr()
```

```python
plt.figure(figsize=(4, 2))
sns.heatmap(vendor_df[["Quantity", "Dollars", "Freight"]].corr(), annot=True)
plt.show()

plt.scatter(vendor_df["Quantity"], vendor_df["Freight"], color="#f57a55")
plt.scatter(vendor_df["Dollars"], vendor_df["Freight"], color="#7f1e5a")
plt.legend(["Quantity", "Dollars"])
plt.ylabel("Freight Cost")
plt.show()
```

```python
vendor_df["freight_per_unit"] = vendor_df["Freight"] / vendor_df["Quantity"]
```

```python
low_quantity = vendor_df["Quantity"].quantile(0.25)
high_quantity = vendor_df["Quantity"].quantile(0.75)

print(f" low quantity: {low_quantity} and high quantity: {high_quantity}")
```

```python
vendor_df.loc[vendor_df["Quantity"] < low_quantity, "freight_per_unit"].mean()
```

```python
vendor_df.loc[vendor_df["Quantity"] > high_quantity, "freight_per_unit"].mean()
```

```python
X = vendor_df[["Quantity", "Dollars"]]
Y = vendor_df[["Freight"]]
```

```python
vendor_df.describe().round()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.20, random_state=11
)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, root_mean_squared_error, r2_score
```

```python
model1 = LinearRegression()
model1.fit(X_train, Y_train)

model2 = DecisionTreeRegressor(max_depth=10, random_state=11)
model2.fit(X_train, Y_train)

model3 = RandomForestRegressor(max_depth=7, random_state=11)
model3.fit(X_train, Y_train)
```

```python
def evaluate_model(model, X_test, Y_test, model_name):
    preds = model.predict(X_test)

    mae = mean_absolute_error(Y_test, preds)
    rmse = root_mean_squared_error(Y_test, preds)
    r2 = r2_score(Y_test, preds) * 100

    print(f"\n{model_name} Performance:")
    print(f"MAE  : {mae:.2f}")
    print(f"RMSE : {rmse:.2f}")
    print(f"R²   : {r2:.2f}%")
```

```python
evaluate_model(model1, X_test, Y_test, "Linear Regresion")
evaluate_model(model2, X_test, Y_test, "Decision Tree Regresion")
evaluate_model(model3, X_test, Y_test, "Random Forest Regresion")
```

```python
preds = model1.predict(X_test)

plt.figure(figsize=(6, 4))
plt.scatter(Y_test, preds, alpha=0.6)

plt.xlabel("Actual Freight Cost")
plt.ylabel("Predicted Freight Cost")
plt.title("Actual vs Predicted Freight Cost")
plt.show()
```

```python
preds = model1.predict(X_test)


sorted_idx = X_test["Quantity"].argsort()

plt.figure(figsize=(8, 5))


plt.scatter(X_test["Quantity"], Y_test, label="Actual", alpha=0.6)


plt.plot(
    X_test["Quantity"].iloc[sorted_idx],
    preds[sorted_idx],
    color="red",
    label="Predicted",
    linewidth=2,
)

plt.xlabel("Quantity")
plt.ylabel("Freight Cost")
plt.title("Actual vs Predicted Freight Cost")
plt.legend()
plt.show()
```
